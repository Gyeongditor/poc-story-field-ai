```python
from openai import OpenAI
import dotenv
import os
```

```python
dotenv.load_dotenv()
password = os.getenv('GPT_API_KEY')
#print(password)
client = OpenAI(api_key=password)
```

```python
with open("test.txt", "r", encoding="utf-8") as f:
    user_story = f.read().strip()
```

```python
prompt = f"""
당신은 창의적인 동화 작가입니다.

아래 실제 이야기를 바탕으로, 동화의 줄거리를 만들어 주세요.
- 5페이지 분량 (A4 기준, 전체 약 1200~1500자 정도)
- 페이지 별 이야기를 써줘 ex) 1페이지: , 2페이지:
[이야기 내용]
\"\"\"{user_story}\"\"\"

[동화 줄거리 시작]
"""
```

```python
response = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "user", "content": prompt}
    ],
    temperature=0.8,
    max_tokens=1500
)
```

```python
story_summary = response.choices[0].message.content
print("\n=== 동화 줄거리 ===\n")
print(story_summary)
```

```python
with open("generated_story.txt", "w", encoding="utf-8") as f:
    f.write(story_summary)
```
