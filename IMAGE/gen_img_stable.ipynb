```python
import os
import dotenv
```

```python
dotenv.load_dotenv()
password = os.getenv('HUG_PASS')
```

```python
def extract_pages_from_text(text):
    pages = {}
    current_page = None
    lines = text.strip().split("\n")
    for line in lines:
        line = line.strip()
        if "페이지:" in line:
            current_page = line.replace("페이지:", "").strip()
            pages[current_page] = ""
        elif current_page:
            pages[current_page] += line + " "
    return pages
```

```python
def eng_extract_pages_from_text(text):
    pages = {}
    current_page = None
    lines = text.strip().split("\n")
    for line in lines:
        line = line.strip()
        if "Page:" in line:
            current_page = line.replace("Page:", "").strip()
            pages[current_page] = ""
        elif current_page:
            pages[current_page] += line + " "
    return pages
```

```python
def make_prompt(text):
    return f"storybook illustration, fairytale style, colorful, detailed, for kids, {text.strip()}"
```

```python
from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "CompVis/stable-diffusion-v1-4",
    torch_dtype=torch.float16,
    revision="fp16",
    use_auth_token=password
).to("cuda")
```

```python
def generate_page_images(pages, output_dir="generated_images"):
    os.makedirs(output_dir, exist_ok=True)

    for page, content in pages.items():
        prompt = make_prompt(content)
        print(f"{page} 생성 중: {prompt}")
        image = pipe(prompt).images[0]
        image.save(os.path.join(output_dir, f"page_{page}.png"))
        print(f"저장됨: page_{page}.png")
```

```python
def eng_generate_page_images(pages, output_dir="eng_generated_images"):
    os.makedirs(output_dir, exist_ok=True)

    for page, content in pages.items():
        prompt = make_prompt(content)
        print(f"{page} 생성 중: {prompt}")
        image = pipe(prompt).images[0]
        image.save(os.path.join(output_dir, f"eng_page_{page}.png"))
        print(f"저장됨: eng_page_{page}.png")
```

```python
def read_generated_story(filepath="generated_story.txt"):
    with open(filepath, "r", encoding="utf-8") as f:
        return f.read()
```

```python
def read_generated_eng_story(filepath="generated_story_eng.txt"):
    with open(filepath, "r", encoding="utf-8") as f:
        return f.read()
```

```python
story_text = read_generated_story("generated_story.txt")

# 페이지별 추출
pages = extract_pages_from_text(story_text)

# 이미지 생성
generate_page_images(pages)
```

```python
story_text = read_generated_eng_story("generated_story_eng.txt")

# 페이지별 추출
pages = eng_extract_pages_from_text(story_text)

# 이미지 생성
eng_generate_page_images(pages)
```
