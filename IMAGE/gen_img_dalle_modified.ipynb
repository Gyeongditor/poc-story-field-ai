```python
import openai
import os
from PIL import Image
import requests
from io import BytesIO
import dotenv
```

```python
dotenv.load_dotenv()
password = os.getenv('GPT_API_KEY')
```

```python
story_file = "generated_story.txt"
output_dir = "dalle_generated_images"
os.makedirs(output_dir, exist_ok=True)
```

```python
with open(story_file, "r", encoding="utf-8") as f:
    lines = [line.strip() for line in f if line.strip()]
```

```python
page_prompts = {}
current_page = None

for line in lines:
    if "페이지:" in line:
        current_page = line.replace("페이지:", "").strip()
    elif current_page:
        page_prompts[current_page] = line
        current_page = None
```

```python
for page, prompt in page_prompts.items():
    print(f"📘 {page} → {prompt}")

    response = openai.Image.create(
        prompt=prompt,
        n=1,
        size="512x512"
    )
```

```python
# OpenAI API 키 설정
openai.api_key = password  # <- 이미 .env에서 불러온 값 사용

for page, prompt in page_prompts.items():
    print(f"📘 {page} → {prompt}")

    response = openai.Image.create(
        prompt=prompt,
        n=1,
        size="512x512"
    )

    image_url = response["data"][0]["url"]
    image_response = requests.get(image_url)
    image = Image.open(BytesIO(image_response.content))

    filename = f"page_{page.zfill(2)}.png"
    save_path = os.path.join(output_dir, filename)
    image.save(save_path)

    print(f"✅ Saved {save_path}")
```
