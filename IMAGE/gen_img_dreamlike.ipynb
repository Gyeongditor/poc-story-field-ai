```python
from diffusers import StableDiffusionPipeline
import torch
```

```python
model_id = "dreamlike-art/dreamlike-diffusion-1.0"
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to("cuda")
```

```python
story_file = "story_eng.txt"
with open(story_file, "r", encoding="utf-8") as f:
    content = f.read()
```

```python
pages = content.strip().split("\n\n")
for page in pages:
    print(page)
    print('')
```

```python
output_dir = "./dreamlike_generated_images"
import os
os.makedirs(output_dir, exist_ok=True)
```

```python
negative_prompt = (
    "blurry, low quality, distorted, extra limbs, cropped, watermark, text, "
    "ugly, deformed, lowres, bad anatomy"
)
```

```python
base_style = (
    "children's book illustration style,"
    "watercolor and gouache texture, soft pastel colors, "
    "warm lighting, consistent character design, whimsical atmosphere,"
    "without any text in the images"
)
```

```python
for idx, page in enumerate(pages, start=1):
    description = page.strip()
    prompt = description + base_style

    print(f"[INFO] Page {idx} 이미지 생성 중...")
    image = pipe(prompt, negative_prompt=negative_prompt).images[0]
    image_path = os.path.join(output_dir, f"page_{idx}.png")
    image.save(image_path)
    print(f"[INFO] {image_path} 저장 완료")

print("[INFO] 모든 페이지 이미지 생성 완료.")
```
